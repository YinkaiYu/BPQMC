# Small Triangular HMC Stage Report

Set `REPORT_DIR` to the output directory produced by `render_small_hmc_stage_report.py`.


```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, Markdown, display

REPORT_DIR = Path("data/triangular_hmc_small_benchmark/stage_report")
cases = pd.read_csv(REPORT_DIR / "stage_cases.csv")
observables = pd.read_csv(REPORT_DIR / "stage_observables.csv")
samples = pd.read_csv(REPORT_DIR / "stage_samples.csv")
cases.sort_values(["Nbos", "U2"])
```

```python
display(Markdown((REPORT_DIR / "report.md").read_text(encoding="utf-8")))
```

```python
display(Image(filename=str(REPORT_DIR / "overview.png")))
```

```python
for obs_name in ["doubleOcc", "squareOcc", "nearestOcc", "IPR", "PF_Gamma", "kinetic"]:
    path = REPORT_DIR / f"observable_{obs_name}.png"
    if path.exists():
        display(Markdown(f"## {obs_name}"))
        display(Image(filename=str(path)))
```

```python
for case_name in cases.sort_values(["Nbos", "U2"])["name"]:
    trace_path = REPORT_DIR / f"trace_{case_name}.png"
    tune_path = REPORT_DIR / f"tune_{case_name}.png"
    display(Markdown(f"## {case_name}"))
    if trace_path.exists():
        display(Image(filename=str(trace_path)))
    if tune_path.exists():
        display(Image(filename=str(tune_path)))
```
